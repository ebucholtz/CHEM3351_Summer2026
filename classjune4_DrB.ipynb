```python
# setting up the environment for RDKit visualization in Jupyter notebooks
from rdkit import Chem  
from rdkit.Chem.Draw import IPythonConsole
from rdkit.Chem import Draw
IPythonConsole.ipython_useSVG=True  #< set this to False if you want PNGs instead of SVGs
```

```python
# slide 6
target = Chem.MolFromSmiles("CC(=O)OC")   # methyl acetate
query = Chem.MolFromSmarts("[CX3](=O)[OX2][#6]")  # ester pattern

result = target.HasSubstructMatch(query)
print(result)
# target is a mol file and with from rdkit.Chem.Draw import IPythonConsole we can directly diplay the structure
target
```

```python
# slide 7
# create a dictionary of molecules with SMILES strings
smiles_dict = {
    "ethanol": "CCO",
    "acetic acid": "CC(=O)O",
    "methyl acetate": "CC(=O)OC",
    "acetamide": "CC(=O)N",
}
```

```python
# Convert SMILES strings in dictionary to RDKit mol objects into a new mol object dictionary
mol_dict = {}

for name, smi in smiles_dict.items():
    mol = Chem.MolFromSmiles(smi)
    mol_dict[name] = mol

Draw.MolsToGridImage(
    list(mol_dict.values()),
    legends=list(mol_dict.keys()),
    molsPerRow=5
)
```

```python
# slide 7
# do we have an ester?
ester_query = Chem.MolFromSmarts("[CX3](=O)[OX2][#6]")  #set a SMARTS query for an ester

for name, mol in mol_dict.items():
    target = mol   
    query = Chem.MolFromSmarts("[CX3](=O)[OX2][#6]")  # ester pattern
    print(name,target.HasSubstructMatch(query),sep=": ")
```

```python
#slide 8
# too broad a query
broad_query = Chem.MolFromSmarts("CO")  #set a SMARTS query for a Carbon-Oxygen bond

for name, mol in mol_dict.items():   #for each item in the mol dictionary (mol_dict)
    target = mol   
    print(name,target.HasSubstructMatch(broad_query),sep=": ")  #check substructure matches and print with a separator of a colon and space
```

```python
#slide 9
# does are list of molecules contain any nitrogens?
nitrogen_query = Chem.MolFromSmarts("[#7]")  #set a SMARTS query for a nitrogen

for name, mol in mol_dict.items():   #for each item in the mol dictionary (mol_dict)
    target = mol   
    print(name,target.HasSubstructMatch(nitrogen_query),sep=": ")  #check substructure matches and print with a separator of a colon and space
```

```python
#slide 12
examples = {"pyridine": "c1ccncc1",
            "aniline": "Nc1ccccc1",
            "triethylamine": "CCN(CC)CC",
            "imidazole": "c1ncc[nH]1"
           }
# Convert SMILES strings in dictionary to RDKit mol objects into a new mol object dictionary
mol_dict2 = {}

for name, smi in examples.items():
    mol = Chem.MolFromSmiles(smi)
    mol_dict2[name] = mol

aromatic_n_query = Chem.MolFromSmarts("[n]")
aliphatic_n_query = Chem.MolFromSmarts("[N]")
any_n_query = Chem.MolFromSmarts("[#7]")


print(
    f"{'Name':<15}"
    f"{'aromatic N':<15}"
    f"{'aliphatic N':<15}"
    f"{'any N':<10}"
)

for name, mol in mol_dict2.items():   #for each item in the mol dictionary (mol_dict)
        print(
        f"{name:<15}"
        f"{str(mol.HasSubstructMatch(aromatic_n_query)):<15}"
        f"{str(mol.HasSubstructMatch(aliphatic_n_query)):<15}"
        f"{str(mol.HasSubstructMatch(any_n_query)):<10}"
    )

Draw.MolsToGridImage(
    list(mol_dict2.values()),
    legends=list(mol_dict2.keys()),
    molsPerRow=5
)
```

```python
#slide 19, 20, 21
examples = {"cyclohexane": "C1CCCCC1",
            "hexane": "CCCCCC",
            "benzene": "c1ccccc1",
            "furan": "o1cccc1",
            "pyridine": "c1ccncc1",
            "thiophene": "s1cccc1",
           }
# Convert SMILES strings in dictionary to RDKit mol objects into a new mol object dictionary
mol_dict7 = {}

for name, smi in examples.items():
    mol = Chem.MolFromSmiles(smi)
    mol_dict7[name] = mol

ring = Chem.MolFromSmarts("[r]")
ring5 = Chem.MolFromSmarts("[r5]")
no_ring = Chem.MolFromSmarts("[!r]")
aromatic_o= Chem.MolFromSmarts("[o]")
ring_heteroatom = Chem.MolFromSmarts("[#7,#8,#16;r]")

print(
    f"{'Name':<25}"
    f"{'ring':<15}"
    f"{'5 member ring':<15}"
    f"{'no ring':<15}"
    f"{'aromatic oxygen':<20}"
    f"{'ring heteroatom':<15}"
)

for name, mol in mol_dict7.items():   #for each item in the mol dictionary (mol_dict)
        print(
        f"{name:<25}"
        f"{str(mol.HasSubstructMatch(ring)):<15}"
        f"{str(mol.HasSubstructMatch(ring5)):<15}"
        f"{str(mol.HasSubstructMatch(no_ring)):<15}"
        f"{str(mol.HasSubstructMatch(aromatic_o)):<20}"
        f"{str(mol.HasSubstructMatch(ring_heteroatom)):<15}"
    )

Draw.MolsToGridImage(
    list(mol_dict7.values()),
    legends=list(mol_dict7.keys()),
    molsPerRow=5
)
```
